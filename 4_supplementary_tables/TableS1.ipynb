```r
library(readr)
```

## Load data

```r
# WGS IDs
wgsIDs <- colnames(read_tsv("covariates.ROSMAP_NIA_WGS.leftnorm.bcftools_qc.plink_qc.pQTL.plink_qc.prune.pca.gz", show_col_types = FALSE))
wgsIDs <- wgsIDs[-1]
stopifnot(length(wgsIDs) == 316)
```

```r
# Mappings WGS IDs to ROSMAP project IDs
mapping <- read_csv("rosmap/wgs/ampad/experimentalData/sampleSheet_1200.csv", show_col_types = FALSE)
stopifnot(all(wgsIDs %in% mapping$SampleID))
projIDs <- mapping$ProjID[mapping$SampleID %in% wgsIDs]
```

```r
# ROSMAP variables
rosmap <- read_csv("dataset_707_cross-sectional_03-17-2026.csv", show_col_types = FALSE)
stopifnot(all(projIDs %in% rosmap$projid))
rosmap <- rosmap[rosmap$projid %in% projIDs, ]
```

## Supplementary Table S1: Cohort characteristics

### Demographics

```r
# Age at death, years, mean (sd)
mean(rosmap$age_death); sd(rosmap$age_death)
```

```r
# Female sex, n (%)
nrow(rosmap) - sum(rosmap$msex); (nrow(rosmap) - sum(rosmap$msex)) / nrow(rosmap)
```

```r
# Education, years, mean (sd)
mean(rosmap$educ); sd(rosmap$educ)
```

### Clinical measures

```r
# Cognitive status at death
# No cognitive impairment (cogdx == 1), n (%)
sum(rosmap$cogdx == 1); sum(rosmap$cogdx == 1) / nrow(rosmap)
```

```r
# Mild cognitive impairment MCI (cogdx 2-3), n (%)
sum(rosmap$cogdx %in% c(2, 3)); sum(rosmap$cogdx %in% c(2, 3)) / nrow(rosmap)
```

```r
# Dementia (cogdx 4-6), n (%)
sum(rosmap$cogdx %in% c(4, 5, 6)); sum(rosmap$cogdx %in% c(4, 5, 6)) / nrow(rosmap)
```

### Postmortem indices

```r
# Postmortem interval, hours, mean (sd)
mean(rosmap$pmi); sd(rosmap$pmi)
```

### Neurodegenerative pathologies

```r
# Lewy body pathology beyond nigra (limbic/neocortical, lewydx_st4 2-3), n (%)
# Consistent with TDP-43 definition: restricted to advanced (limbic/neocortical) stages, not any-stage.
sum(rosmap$lewydx_st4 %in% c(2, 3), na.rm = TRUE); sum(rosmap$lewydx_st4 %in% c(2, 3), na.rm = TRUE) / sum(!is.na(rosmap$lewydx_st4))
```

```r
# Alzheimer disease (NIA-AA), n (%)
sum(rosmap$ad_adnc == 1); sum(rosmap$ad_adnc == 1) / nrow(rosmap)
```

```r
# TDP-43, n (%)
sum(rosmap$tdp_st4 %in% c(2, 3), na.rm = TRUE); sum(rosmap$tdp_st4 %in% c(2, 3), na.rm = TRUE) / sum(!is.na(rosmap$tdp_st4))
```

```r
# Hippocampal sclerosis, n (%)
sum(rosmap$hip_scl_yn_mid == 1); sum(rosmap$hip_scl_yn_mid == 1) / nrow(rosmap)
```

### Cerebrovascular pathologies

```r
# Macroinfarcts present, n (%)
sum(rosmap$ci_num2_gct == 1); sum(rosmap$ci_num2_gct == 1) / nrow(rosmap)
```

```r
# Microinfarcts present, n (%)
sum(rosmap$ci_num2_mct == 1); sum(rosmap$ci_num2_mct == 1) / nrow(rosmap)
```

```r
# Atherosclerosis (moderate/severe), n (%)
sum(rosmap$cvda_4gp2 %in% c(2, 3), na.rm = TRUE); sum(rosmap$cvda_4gp2 %in% c(2, 3), na.rm = TRUE) / sum(!is.na(rosmap$cvda_4gp2))
```

```r
# Arteriolosclerosis (moderate/severe), n (%)
sum(rosmap$arteriol_scler %in% c(2, 3), na.rm = TRUE); sum(rosmap$arteriol_scler %in% c(2, 3), na.rm = TRUE) / sum(!is.na(rosmap$arteriol_scler))
```

```r
# Cerebral amyloid angiopathy (moderate/severe), n (%)
sum(rosmap$caa_4gp %in% c(2, 3), na.rm = TRUE); sum(rosmap$caa_4gp %in% c(2, 3), na.rm = TRUE) / sum(!is.na(rosmap$caa_4gp))
```

```r
# Export Table S1
n <- nrow(rosmap)
tableS1 <- data.frame(
  Category = c(
    "Demographics", "", "", "",
    "Clinical measures", "", "", "",
    "Postmortem indices", "",
    "Neurodegenerative pathologies", "", "", "",
    "Cerebrovascular pathologies", "", "", "", ""
  ),
  Variable = c(
    "Age at death, years", "Female sex", "Education, years", "",
    "No cognitive impairment", "Mild cognitive impairment (MCI)", "Dementia", "",
    "Postmortem interval, hours", "",
    "Lewy body disease (limbic or neocortical type)", "Alzheimer disease (NIA-AA)", "TDP-43 present (limbic or neocortical)", "Hippocampal sclerosis",
    "Macroinfarcts present", "Microinfarcts present", "Atherosclerosis (moderate/severe)", "Arteriolosclerosis (moderate/severe)", "Cerebral amyloid angiopathy (moderate/severe)"
  ),
  Statistic = c(
    "mean (SD)", "n (%)", "mean (SD)", "",
    "n (%)", "n (%)", "n (%)", "",
    "mean (SD)", "",
    "n (%)", "n (%)", "n (%)", "n (%)",
    "n (%)", "n (%)", "n (%)", "n (%)", "n (%)"
  ),
  Value = c(
    sprintf("%.2f (%.2f)", mean(rosmap$age_death), sd(rosmap$age_death)),
    sprintf("%d (%.1f%%)", n - sum(rosmap$msex), 100 * (n - sum(rosmap$msex)) / n),
    sprintf("%.2f (%.2f)", mean(rosmap$educ), sd(rosmap$educ)),
    "",
    sprintf("%d (%.1f%%)", sum(rosmap$cogdx == 1), 100 * sum(rosmap$cogdx == 1) / n),
    sprintf("%d (%.1f%%)", sum(rosmap$cogdx %in% c(2,3)), 100 * sum(rosmap$cogdx %in% c(2,3)) / n),
    sprintf("%d (%.1f%%)", sum(rosmap$cogdx %in% c(4,5,6)), 100 * sum(rosmap$cogdx %in% c(4,5,6)) / n),
    "",
    sprintf("%.2f (%.2f)", mean(rosmap$pmi), sd(rosmap$pmi)),
    "",
    sprintf("%d (%.1f%%)", sum(rosmap$lewydx_st4 %in% c(2,3), na.rm=TRUE), 100 * sum(rosmap$lewydx_st4 %in% c(2,3), na.rm=TRUE) / sum(!is.na(rosmap$lewydx_st4))),
    sprintf("%d (%.1f%%)", sum(rosmap$ad_adnc == 1), 100 * sum(rosmap$ad_adnc == 1) / n),
    sprintf("%d (%.1f%%)", sum(rosmap$tdp_st4 %in% c(2,3), na.rm=TRUE), 100 * sum(rosmap$tdp_st4 %in% c(2,3), na.rm=TRUE) / sum(!is.na(rosmap$tdp_st4))),
    sprintf("%d (%.1f%%)", sum(rosmap$hip_scl_yn_mid == 1), 100 * sum(rosmap$hip_scl_yn_mid == 1) / n),
    sprintf("%d (%.1f%%)", sum(rosmap$ci_num2_gct == 1), 100 * sum(rosmap$ci_num2_gct == 1) / n),
    sprintf("%d (%.1f%%)", sum(rosmap$ci_num2_mct == 1), 100 * sum(rosmap$ci_num2_mct == 1) / n),
    sprintf("%d (%.1f%%)", sum(rosmap$cvda_4gp2 %in% c(2,3), na.rm=TRUE), 100 * sum(rosmap$cvda_4gp2 %in% c(2,3), na.rm=TRUE) / sum(!is.na(rosmap$cvda_4gp2))),
    sprintf("%d (%.1f%%)", sum(rosmap$arteriol_scler %in% c(2,3), na.rm=TRUE), 100 * sum(rosmap$arteriol_scler %in% c(2,3), na.rm=TRUE) / sum(!is.na(rosmap$arteriol_scler))),
    sprintf("%d (%.1f%%)", sum(rosmap$caa_4gp %in% c(2,3), na.rm=TRUE), 100 * sum(rosmap$caa_4gp %in% c(2,3), na.rm=TRUE) / sum(!is.na(rosmap$caa_4gp)))
  )
)
write_csv(tableS1, "TableS1.csv")
```

```r
# Source-level check: Table S1 is built entirely from WGS sample IDs (covariate file
# colnames) and ROSMAP phenotype variables (demographics/clinical/pathology) -- it never
# loads gpAnnotation or any glycopeptiform-level object, so the w_sole_exclude_gpids (54
# sole-modification Trp C-mannosylation glycopeptiforms mislabeled as glycation) cannot
# appear in its inputs or output. Verified directly rather than assumed.
w_sole_exclude_gpids <- readLines("../2_main_figures/Figure1/data/w_sole_exclude_gpids.txt")
stopifnot(length(w_sole_exclude_gpids) == 54)
qc_remove_gpids <- readLines("../2_main_figures/Figure1/data/qc_remove_gpids.txt")
excluded_gpids <- c(qc_remove_gpids, w_sole_exclude_gpids)
src_strings <- c(wgsIDs, mapping$SampleID, mapping$ProjID, colnames(rosmap), unlist(tableS1))
hits <- intersect(unique(as.character(src_strings)), excluded_gpids)
stopifnot(length(hits) == 0)
cat("No affected gpIDs found in TableS1 source objects (sample/phenotype data only, no glycopeptiform-level input).\n")
```

```r
# Coding-variant check (v202609): 21 glycopeptiforms dropped after the coding-variant screen must not enter any QTL-level object of this table
coding_drop_gpids <- readLines('../2_main_figures/Figure1/data/coding_variant_drop_gpids.txt')
stopifnot(length(coding_drop_gpids) == 21L, !anyDuplicated(coding_drop_gpids))
# Table S1 is built only from sample IDs and ROSMAP phenotype variables (no glycopeptiform-level input), so no gpID can appear
src_strings <- c(wgsIDs, mapping$SampleID, mapping$ProjID, colnames(rosmap), unlist(tableS1))
hits <- intersect(unique(as.character(src_strings)), coding_drop_gpids)
stopifnot(length(hits) == 0)
cat('No coding-variant-dropped gpIDs found in TableS1 source objects (sample/phenotype data only).\n')
```
